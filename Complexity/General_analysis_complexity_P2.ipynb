# Análisis A priori de la función `prod`

Con base a al definición de la función `prod`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.


```python

def prod(a: int, b: int) -> int:
    if b == 1: # O(1)
        return a 
    else:
        return a + prod(a, b - 1) # T(b - 1)
    

```

Supongamos que el mejor caso ocurre cuando `a > b`, lo que sería $\Theta(b)$. Por otro lado, el peor caso sería cuando `a >= b` lo que implicaría una complejida de $\Theta(b)$, y como observamos queda en función del valor de `b` en todos los casos, entonces no se peuda hacer la distinción de mejor y peor caso. 

Sea $T(b)$ el número de invocaciones de la función.

$$

T(b) = T(b - 1) + 1 \\
= [T(b - 2) + 1] + 1 \\
= T(b - 2) + 2 \\
= [T(b - 3) + 1] + 2
= T(b - 3)+ 3 \\
. \\
. \\
. \\
= T(b - i) + i


$$

Si $b - i = 1$ , entonces $i = b - 1$  ,

$$
= T(b - (b - 1)) + (b - 1)
= T(1) + (b - 1)
\therefore T(b) \in \Theta(b)
$$

Cabe mencionar que se propuso $i = b - 1$ dado el caso base cuando $b = 1$. En general, la función será invocada exactamente `b - 1` veces





````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, N, label=r"$\Theta(n)$")

plt.title(f"Complejidad Producto Recursivo")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A posteriori de la función `prod`

## Implementación

La función `prod` tiene el objetivo de realizar el producto de dos números recursivamente al restar una unidad al número `b`en cada invocación hasta que `b`sea 1, es decir, se va a sumar `a` la cantidad de `b`.

A continuación, se muestra la implementación.

````python
def prod(a: int, b: int) -> int:
    if b == 1:
        return a
    else:
        return a + prod(a, b - 1)

a = 7
b = 7

print("Resultado prod: ", prod(a, b))

size_data = [
    10,
    50,
    100,
    500,
    1000,
]


# ==========================================================
# MEDICIÓN EXPERIMENTAL
# ==========================================================

times = []

REPETITIONS = 1000

for n in size_data:

    # Peor caso aproximado:
    # prod(n,1)
    x = n
    y = n

    start = time.perf_counter()

    for _ in range(REPETITIONS):
        prod(x, y)

    end = time.perf_counter()

    # promedio en milisegundos
    curr_time = (
        (end - start) / REPETITIONS
    ) * 1000

    print(
        f"n={n} -> "
        f"{curr_time:.8f} ms"
    )

    times.append(curr_time)


# ==========================================================
# COMPLEJIDAD TEÓRICA Θ(n)
# ==========================================================

theoretical_raw = size_data

# Escalamiento para comparar tendencia
scale_factor = (
    times[-1] /
    theoretical_raw[-1]
)

theoretical = [
    x * scale_factor
    for x in theoretical_raw
]


# ==========================================================
# RATIO
# ==========================================================

ratio = [
    times[i] / size_data[i]
    for i in range(len(size_data))
]


# ==========================================================
# GRÁFICAS
# ==========================================================

fig, axes = plt.subplots(1,2,figsize=(15, 5))
fig.suptitle(
    "Análisis A Posteriori — Producto Recursivo",
    fontsize=14,
    y=1.02
)

# ----------------------------------------------------------
# 1. Tiempo real vs Teórico
# ----------------------------------------------------------

axes[0].plot(
    size_data,
    times,
    "o-",
    color="#E24B4A",
    linewidth=2,
    label="Tiempo real"
)
axes[0].plot(
    size_data,
    theoretical,
    "--",
    color="#378ADD",
    linewidth=2,
    label=r"$\Theta(n)$ teórico"
)
axes[0].set_xscale("log")
axes[0].set_yscale("log")
axes[0].set_title("Tiempo real vs Teórico")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Tiempo promedio (ms)")
axes[0].legend()
axes[0].grid(alpha=0.4)

# ----------------------------------------------------------
# 2. Ratio
# ----------------------------------------------------------

axes[1].plot(
    size_data,
    ratio,
    "o-",
    color="#1D9E75",
    linewidth=2
)
axes[1].set_xscale("log")
axes[1].set_title(r"Ratio $T(n)/n$")
axes[1].set_xlabel("n")
axes[1].set_ylabel(r"$T(n)/n$")

axes[1].grid(alpha=0.4)

# ==========================================================

filename = "prod_aposteriori.png"
plt.tight_layout()
plt.savefig(
    f"IMG/{filename}",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
print(f"\nGráfica guardada como {filename}")
````

# Análisis A priori de la función `binario`

Con base a al definición de la función `binario`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.


```python

def binario(n: int) -> None:
    if n < 2: # O(1)
        print(n, end=" ")
    else:
        binario(n//2) # T(n/2)
        print(n%2, end=" ") # O(1)

```
Para la función `binario`, tenemos que en cada invocación nuestra entrada se aplica la división entera y se detiene hasta que `n < 2`. En este caso, no se puede hacer una distinción del mejor caso, porque el número de invoaciones esta sujeto a $n/2 \in Z^+$.

Ahora bien, se puede construir la ecuación de recursividad de al siguiente forma:


Sea $T(n)$ el número de invocaciones de la función.

$$
 T(n) = T(n/2) + C
$$

Aplicando el Teorema Maestro con la forma  $T(n) = aT(n/b) + f(n)$, donde $a = 1$ , $b = 2$ y $f(n) = C$. Obtenemos lo siguiente:

$$
n^{log_{b}a}  = n^{log_{2}1}  = 1
$$

luego:

$$

f(n) = C  = \Theta(1) = \Theta(n^{log_{2}1}) = \Theta(1) \\
T(n) = \Theta(n^{log_{2}1} * log(n) \\
\therefore T(n) \in \Theta(log n) 

$$




````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, np.log2(N), label=r"$O(\log_{2}(n))$")

plt.title(f"Complejidad Función Binario")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A priori de la función `i`

Con base a al definición de la función `i`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.


```python

def foo_i(n: int) -> None:
    if n <= 1: # O(1)
        return
    
    for i in range(n): # O(n)
        print("*", end="")
    print()
    foo_i(int(0.8*n)) # T(4/5*n)

```

La función `foo_i`, constuye una piramide invertida con asteriscos. En este caso no se puede hacer un distinción del mejor caso, aunque su caso base es cuando $n \leq 1$. Por lo tanto, podemos construir la ecuación recursiva.

$$
T(n) = T(4n / 5) + n 
$$

Aplicando el Teorema Maestro con la forma  $T(n) = aT(n/b) + f(n)$, donde $a = 1$ , $b = 5/4$ y $f(n) = n$. Obtenemos lo siguiente:

$$

n^{log_{b}(a)}  = n^{log_{5/4}(1)} = n^{0}  = 1
$$

Ahora bien, se aplica el caso 3 dado que $f(n)$ domina sobre $n^{log_{b}(a)}$, es decir
$$
f(n) = \Omega(n^{log_{b}(a) + \epsilon}) , \quad \epsilon > 0
$$

Entonces,

$$
f(n) = n = n^{0 + \epsilon}, \quad \epsilon = 1 \\

\therefore T(n) \in \Theta(n)
$$








````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, N, label=r"$O(n)$")

plt.title(f"Complejidad Función i")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A priori de la función `j`

Con base a al definición de la función `j`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.


```python
def foo_j(n: int) -> None:
    if n < 2: # O(1)
        return
    else:
        counter = 0 # O(1)
        for i in range(8): # 8T(n/2)
            foo_j(n//2)
        
        for i in range(n**3): # T(n^3) 
            counter += 1
```

Como podemos observar, tenemos el bloque else donde tenemos que determinar su complejidad. En general, el bloque es $8T(n/2) + T(n^3) + C$, por lo que podemos construir la ecuación recursiva

$$

T(n) = 8T(n/2) + T(n^3)

$$

Aplicando el Teorema Maestro con la forma  $T(n) = aT(n/b) + f(n)$, donde $a = 8$ , $b = 2$ y $f(n) = n^3$. Obtenemos lo siguiente:

$$

n^{\log_{b}(a)} = n^{\log_{2}(8)} = n^{3}

$$

Ahora bien, se aplica el caso 2 dado que $n^{log_{b}(a)}$ colapsa directamente al valor de $f(n)$, es decir

$$
f(n) = n^3 = \Theta(n^{log_{b}(a)}) = n^3 \\

$$

Entonces,

$$
\Theta(n^{log_{2}(8)} \log(n)) \\
\therefore T(n) \in \Theta(n^3 \log_{2}(n))
$$








````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, (N**3) * np.log2(N), label=r"$\Theta(n^3 \log_{2}(n))$")

plt.title(f"Complejidad Función j")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A priori de la función `k`

Con base a al definición de la función `k`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.


```python

def foo_k(n: int) -> None:
    i = 1
    while i <= n : # O(n)
        j = 1
        while j <= n: # O(log2 n)
            print(" * ", end="") 
            j *= 2
        print()
        i += 1

```

Como se puede observar, no se puede hacer alguna distinción entre mejor y peor caso. Por lo tanto, utilizaremos la notación $\Theta$.

Ciclo externo, `i` va de 1 hasta n de uno en uno:

$$

1, 2, 3, . . ., n \to \Theta(n)

$$

Ciclo interno, `j` se duplica en cada iteración:

$$

1, 2, 4, 8, ..., 2^k

$$

El ciclo se deteine cuando $2^k > n$, entonces:

$$

2^k = n \to k = \log_{2}(n) \to \Theta(\log_{2}(n))

$$

Por lo tanto, la complejidad es

$$

\therefore T(n) \in \Theta(n \log_{2}(n))

$$


````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, N * np.log2(N), label=r"$\Theta(n \log_{2}(n))$")

plt.title(f"Complejidad Función k")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A posteriori de la función `k`

## Implementación

La función `k` tiene como objetivo impirmir una rectangulo con asteriscos.

        binary(n - 1, A) # T(n - 1)

La función genera todas las cadenas binarias posibles de longitud `n`. Como no hay distinción entre mejor y peor caso, utilizaremos la notación $\Theta$.

$$

= 4[2T(n - 3) + C] + 3C \to 8T(n - 3) + 7C \\
\vdots \\
= 2^i T(n - i) + (2^i - 1)C
$$

Sea $n - i = 0 \implies i = n$, entonces:

$$
= 2^{n}T(0) + (2^n - 1)C \\
\therefore T(n) \in \Theta(2^{n})

La función debe de tener un orden de complejidad lineal.

````python
def checkSideSums(A: list[int]) -> bool:
    right_side = 0
    for i in range(len(A)):
        right_side += A[i]

    left_side = 0
    for i in range(len(A)):
        right_side -= A[i]
        if right_side == left_side:
            return True
        
        left_side += A[i]
    return False
    

# A = [1, 2, 3, 4, 5,6, 7, 8]
A = [1, 2, 3]

print("Resultado: ", checkSideSums(A))
````

Sea $\sqrt{n} = n^{1/2}$ y $n = 2^m \to m = \log{n}$, tenemos:

$$

T(2^m) = T(\sqrt{2^m}) + C = T(2^{m/2}) + C \\

$$

Definimos $S(m) = T(2^{m})$, lo que nos da:

$$

S(m) = S(m/2) + C

$$

Aplicando el Teorema Maestro, donde $a = 1$ , $b = 2$ y $f(m) = C$. Obtenemos lo siguiente:

$$

m^{\log_{b}(a)} = m^{\log_{2}(1)} = 1

$$

Ahora bien, se aplica el caso 2 debido a que $m^{\log_{b}(a)}$ colapsa a 1, entonces :


$$

f(m) = 1 = O(m^{log_{2}(1)}) = 1 = 1 \\

$$

Entonces,

$$
\Theta(\log{m}) \to \Theta(\log{\log{n}}) \\
\therefore T(n) \in \Theta(\log{\log{n}})
$$








````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 50)


plt.plot(N, np.log2(np.log2(N)), label=r"$\Theta(\log{\log{n}})$")

plt.title(f"Complejidad Función F")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

## Análisis A posteriori

A continuación, se muestra su implementación

````python
from math import floor, sqrt

def foo_f(n: int) -> int:
    if n <= 2:
        return 1
    else:
        return foo_f(floor(sqrt(n))) + 1

print("Resultado: ", foo_f(10))

size_data = [
    10,
    20,
    50,
    75,
    95,
    100,
    200,
    500,
    1000,
    50000
]

# ------------------------------------------------------------------ #
#  MEDICIÓN EXPERIMENTAL                                             #
# ------------------------------------------------------------------ #

times = []

REPETITIONS = 10000

for n in size_data:
    
    start = time.perf_counter()
    for _ in range(REPETITIONS):
        foo_f(n)
    end = time.perf_counter()
    # promedio en milisegundos
    curr_time = ((end - start) / REPETITIONS) * 1000
    print(f"n={n} -> {curr_time:.10f} ms")
    times.append(curr_time)

# ------------------------------------------------------------------ #
#  COMPLEJIDAD TEÓRICA                                               #
# ------------------------------------------------------------------ #

theoretical_raw = [np.log2(np.log2(i)) for i in size_data]

# Escalar
scale_factor = times[-1] / theoretical_raw[-1]

theoretical = [
    x * scale_factor
    for x in theoretical_raw
]

# ------------------------------------------------------------------ #
#  RATIO                                                             #
# ------------------------------------------------------------------ #

ratio = [
    times[i] / theoretical_raw[i]
    for i in range(len(size_data))
]

# ------------------------------------------------------------------ #
#  GRÁFICAS                                                          #
# ------------------------------------------------------------------ #

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

fig.suptitle(
    "Análisis A Posteriori — Función F",
    fontsize=14,
    y=1.02
)

# ==========================================================
# 1. Tiempo real vs Teórico
# ==========================================================

axes[0].plot(
    size_data,
    times,
    "o-",
    color="#E24B4A",
    linewidth=2,
    label="Tiempo real"
)

axes[0].plot(
    size_data,
    theoretical,
    "--",
    color="#378ADD",
    linewidth=2,
    label=r"$\Theta(\log{\log{n}})$ teórico"
)

axes[0].set_xscale("log")
axes[0].set_title("Tiempo real vs Teórico")
axes[0].set_xlabel("n (número de elementos)")
axes[0].set_ylabel("Tiempo promedio (ms)")
axes[0].legend()
axes[0].grid(alpha=0.4)

# ==========================================================
# 2. Ratio
# ==========================================================

axes[1].plot(
    size_data,
    ratio,
    "o-",
    color="#1D9E75",
    linewidth=2
)

axes[1].set_xscale("log")
axes[1].set_title(r"Ratio $T(n)/\log{\log{n}}$")
axes[1].set_xlabel("n (número de elementos)")
axes[1].set_ylabel(r"$T(n)/n$")
axes[1].grid(alpha=0.4)

# ------------------------------------------------------------------ #

filename = "foo_f_aposteriori.png"
plt.tight_layout()
plt.savefig(
    f"IMG/{filename}",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
print(f"\nGráfica guardada como {filename}")
````
