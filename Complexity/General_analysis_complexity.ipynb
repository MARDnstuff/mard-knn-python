# Análisis A priori de la función `Búsqueda Binaria`

Con base a al definición de la función `Búsqueda Binaria`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.

Se asume que la entrada esta ordenada de forma ascendente, y sea `T` la función y `n` el tamaño del arreglo.

```python
def binarySearch(arr: list[int], target: int, inf: int, sup: int) -> int: # T(n)
    """
    Busqueda Binaria

    :param arr: arreglo de enteros ordenados ascendentemente
    :param n: entero a buscar
    :param inf: límite inferior
    :param sup: límite superior
    :return: índice del entero a buscar o -1 si no se encuentra
    """
    if inf > sup: # O(1)
        return -1

    mid = (inf + sup) // 2 # O(1)

    if arr[mid] == target:
        return mid
    elif target < arr[mid]: # T(n/2)
        return binarySearch(arr, target, inf, mid - 1)
    else:
        return binarySearch(arr, target, mid + 1, sup)
```

Entonces, se puede observar que el mejor caso ocurre cuando en el primer intento `mid` es igual al elemento que estamos buscando. Ahora bien, el peor caso ocurrira cuando el elemento no se encuentre en el arreglo, y es aquí donde vemos el comportamiento real del algoritmo donde en cada iteración el área de búsqueda se reduce a la mitad, lo que tenemos:

$$

T(n/2) , \quad sea \quad n = 2^{k} = k * log_{2}(n): \\

    T(2^{k}) = T(2^{k - 1}) + C \\
    = [T(2^{k - 2}) + C]+ C \\
    = [T(2^{k - 2}) + C]+ C \\
    = T(2^{k - 2}) +  2C \\
    = [T(2^{k - 3}) + C]+ 2C \\
    = T(2^{k - 3}) +  3C  \\
    . \\
    . \\
    . \\
    = T(2^{k - i}) +  iC \\

Si \quad k - i = 0, \quad entonces \quad k = i  \quad y \quad k = log_{2}(n) \\

= T(2^{0}) + kC \\
= C + kC \\
= C + log_{2}(n) * C
$$

Por lo tanto, tenemos que el peor caso es $O(log_{2}(n))$. En resumen tenemos:

- Mejor caso: $\Omega(1)$
- Peor caso: $O(log_{2}(n))$

NOTA:
- No se tiene $\Theta$ dado que la función no crece al mismo ritmo en sus casos abordados.


````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, np.log2(N), label=r"$O(\log n)$")
plt.plot(N, np.ones(len(N)), label=r"$\Omega(1)$")

plt.title(f"Complejidad Búsqueda Binaria")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A posteriori de la función `Búsqueda Binaria`

## Implementación

La idea de la Búsqueda Binaria es el encontrar un elemento al dentro de un arreglo ordenado ascendemente. En general, se define la mitad del arreglo y se va a comparando la elemento de la mitad con el elemento en cuestión, si el elemento a buscar es mayor que el elemnento en la mitad, entonces reducimos la busqueda al lado derecho del arreglo. En otro coso, nos enfocamos en el lado izquierdo, y esto se repite hasta que los indices que definen el área de busqueda son igual o se cruzan.

Con esta idea en mente, se implementaría una `Búsqueda Binaria`

````python

def binarySearch(arr: list[int], target: int, inf: int, sup: int) -> int:
    """
    Busqueda Binaria

    :param arr: arreglo de enteros ordenados ascendentemente
    :param target: entero a buscar
    :param inf: límite inferior
    :param sup: límite superior
    :return: índice del entero a buscar o -1 si no se encuentra
    """
    if inf > sup:
        return -1

    mid = (inf + sup) // 2

    if arr[mid] == target:
        return mid
    elif target < arr[mid]:
        return binarySearch(arr, target, inf, mid - 1)
    else:
        return binarySearch(arr, target, mid + 1, sup)
    

arr = [1, 2, 3, 4, 5, 6]
n = 7
print("Indice: ", binarySearch(arr, n, 0, len(arr) - 1))
````

````python
import time
import numpy as np
import matplotlib.pyplot as plt

size_data = [
    10,
    100,
    1000,
    5000,
    10000,
    50000,
    100000,
    500000,
    1000000
]

# ------------------------------------------------------------------ #
#  DATASET                                                           #
# ------------------------------------------------------------------ #

test_data = []

for size in size_data:
    sample = []
    for i in range(size):
        sample.append(i)
    test_data.append(sample)

# ------------------------------------------------------------------ #
#  MEDICIÓN EXPERIMENTAL                                             #
# ------------------------------------------------------------------ #

n_size = []
times = []

REPETITIONS = 10000

for test in test_data:
    target = -1
    start = time.perf_counter()
    for _ in range(REPETITIONS):
        binarySearch(test, target, 0, len(test) - 1)
    end = time.perf_counter()
    # promedio en milisegundos
    curr_time = ((end - start) / REPETITIONS) * 1000
    print(f"n={len(test)} -> {curr_time:.10f} ms")
    times.append(curr_time)
    n_size.append(len(test))

# ------------------------------------------------------------------ #
#  COMPLEJIDAD TEÓRICA                                               #
# ------------------------------------------------------------------ #

theoretical_raw = [np.log2(i) for i in size_data]

# Escalar O(log n) para comparar tendencias
scale_factor = times[-1] / theoretical_raw[-1]

theoretical = [
    x * scale_factor
    for x in theoretical_raw
]

# ------------------------------------------------------------------ #
#  RATIO                                                             #
# ------------------------------------------------------------------ #

ratio = [
    times[i] / theoretical_raw[i]
    for i in range(len(size_data))
]

# ------------------------------------------------------------------ #
#  GRÁFICAS                                                          #
# ------------------------------------------------------------------ #

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

fig.suptitle(
    "Análisis A Posteriori — Binary Search",
    fontsize=14,
    y=1.02
)

# ==========================================================
# 1. Tiempo real vs Teórico
# ==========================================================

axes[0].plot(
    size_data,
    times,
    "o-",
    color="#E24B4A",
    linewidth=2,
    label="Tiempo real"
)

axes[0].plot(
    size_data,
    theoretical,
    "--",
    color="#378ADD",
    linewidth=2,
    label=r"$O(\log n)$ teórico"
)

axes[0].set_xscale("log")
axes[0].set_title("Tiempo real vs Teórico")
axes[0].set_xlabel("n (número de elementos)")
axes[0].set_ylabel("Tiempo promedio (ms)")
axes[0].legend()
axes[0].grid(alpha=0.4)

# ==========================================================
# 2. Ratio
# ==========================================================

axes[1].plot(
    size_data,
    ratio,
    "o-",
    color="#1D9E75",
    linewidth=2
)

axes[1].set_xscale("log")
axes[1].set_title(r"Ratio $T(n)/\log_2(n)$")
axes[1].set_xlabel("n (número de elementos)")
axes[1].set_ylabel(r"$T(n)/\log_2(n)$")
axes[1].grid(alpha=0.4)

# ------------------------------------------------------------------ #

plt.tight_layout()

plt.savefig(
    "IMG/binarySearch_aposteriori.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print("\nGráfica guardada como binarySearch_aposteriori.png")
````

# Análisis A priori de la función `Mínimo 1`

Con base a al definición de la función `Mínimo 1`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.

```python

def minimo1(arr: int , ini: int, fin: int) -> int:
    if ini == fin: # O(1)
        return arr[ini]
    else:
        mid = (ini + fin) // 2 # O(1)
        min1 = minimo1(arr, ini, mid) # T(n/2)
        min2 = minimo1(arr, mid + 1, fin) # T(n/2)

        if min1 < min2: # O(1)
            return min1
        else: # O(1)
            return min2
```

Aquí podemos ver que el problema divide en dos el área, pero dado que se busca el minimo en los dos lados, el problema debería colapsar a una complejidad de $O(n)$ simplemente por el hecho de que sigue explorando todos los elementos.

Sin emabargo, eso lo podemos ver claramente al construir la recurrencia:

$$
T(n) = [ T(n/2) + 1 ] + [ T(n/2) + 1 ] \\
T(n) = 2T(n/2) + C
$$

Y vemos que es igual a la forma $T(n) = aT(n/b) + f(n)$, donde podemos aplicar el Teorema maestro. Entonces $a = 2$ , $b = 2$ y $f(n) = C$.

$$

n^{log_{b}(a)}, \quad n^{log_{2}(2)} = n  \\

y \\

f(n) = O(C), como C < n \quad C  \quad es \quad polinomialmente \quad menor \\

$$

Por lo tanto, C no crece más rápido que n ,y eso nos da una complejidad de $\Theta(n)$


````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, N, label=r"$\Theta(n)$")

plt.title(f"Complejidad Mínimo 1")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A posteriori de la función `Mínimo 1`

## Implementación
La idea central del algoritmo es encontrar un medio y buscar el menor de esas mitadas una y otra vez, es un caso típico de la técnia de dividir y conquistar.

A continuación, se muestra su implementación


````python
import time
import random
import numpy as np
import matplotlib.pyplot as plt

def minimo1(arr: int , ini: int, fin: int) -> int:
    if ini == fin:
        return arr[ini]
    else:
        mid = (ini + fin) // 2
        min1 = minimo1(arr, ini, mid)
        min2 = minimo1(arr, mid + 1, fin)

        if min1 < min2:
            return min1
        else:
            return min2

arr = [1, 2, 3, 4, 5, 7, -23, 40]
print("Resultado: ", minimo1(arr, 0, len(arr) - 1))


size_data = [
    10,
    100,
    1000,
    5000,
    10000,
    50000,
    100000,
    500000,
    1000000
]


# ==========================================================
# GENERACIÓN DE DATASETS
# ==========================================================

test_data = []

for size in size_data:
    sample = [
        random.randint(-1000000, 1000000)
        for _ in range(size)
    ]
    test_data.append(sample)


# ==========================================================
# MEDICIÓN EXPERIMENTAL
# ==========================================================

times = []

REPETITIONS = 100

for test in test_data:
    start = time.perf_counter()
    for _ in range(REPETITIONS):
        minimo1(
            test,
            0,
            len(test) - 1
        )
    end = time.perf_counter()
    # promedio en milisegundos
    curr_time = (
        (end - start) / REPETITIONS
    ) * 1000

    print(f"n={len(test)} -> "f"{curr_time:.8f} ms")
    times.append(curr_time)


# ==========================================================
# COMPLEJIDAD TEÓRICA Θ(n)
# ==========================================================

theoretical_raw = size_data

# Escalar para comparar tendencias
scale_factor = (
    times[-1] /
    theoretical_raw[-1]
)

theoretical = [
    x * scale_factor
    for x in theoretical_raw
]


# ==========================================================
# RATIO
# ==========================================================

ratio = [
    times[i] / size_data[i]
    for i in range(len(size_data))
]


# ==========================================================
# GRÁFICAS
# ==========================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(15, 5)
)

fig.suptitle(
    "Análisis A Posteriori — minimo1",
    fontsize=14,
    y=1.02
)

# ----------------------------------------------------------
# 1. Tiempo real vs teórico
# ----------------------------------------------------------

axes[0].plot(
    size_data,
    times,
    "o-",
    color="#E24B4A",
    linewidth=2,
    label="Tiempo real"
)

axes[0].plot(
    size_data,
    theoretical,
    "--",
    color="#378ADD",
    linewidth=2,
    label=r"$\Theta(n)$ teórico"
)

axes[0].set_xscale("log")
axes[0].set_yscale("log")

axes[0].set_title(
    "Tiempo real vs Teórico"
)

axes[0].set_xlabel(
    "n (número de elementos)"
)

axes[0].set_ylabel(
    "Tiempo promedio (ms)"
)

axes[0].legend()

axes[0].grid(alpha=0.4)

# ----------------------------------------------------------
# 2. Ratio
# ----------------------------------------------------------

axes[1].plot(
    size_data,
    ratio,
    "o-",
    color="#1D9E75",
    linewidth=2
)

axes[1].set_xscale("log")

axes[1].set_title(
    r"Ratio $T(n)/n$"
)

axes[1].set_xlabel(
    "n (número de elementos)"
)

axes[1].set_ylabel(
    r"$T(n)/n$"
)

axes[1].grid(alpha=0.4)

# ==========================================================

plt.tight_layout()

plt.savefig(
    "IMG/minimo1_aposteriori.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "\nGráfica guardada como "
    "minimo1_aposteriori.png"
)
````

# Análisis A priori de la función `Mínimo 2`

Con base a al definición de la función `Mínimo 2`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.


```python

def minimo2(A: int , n: int) -> int:
    if n == 0: # O(1)
        return A[0]
    else:
        aux = minimo2(A, n - 1) # T(n - 1)
        if A[n] > aux: # O (1)
            return aux
        else:

```

En esta función es clara la complejidad, dado que esta recursividad fuerza a la función a recorrer cada elemento. Intuitivamente, se compara el elemento actual `A[n]` con el mínimo que se encontro del resto del arreglo y en cada llamada se realiza lo mismo, por lo que un primera conclusión sería tener un complejidad de $O(n)$. 


````python
import numpy as np
import matplotlib.pyplot as plt

N = np.arange(1, 1000)


plt.plot(N, N, label=r"$O(\log n)$")
plt.plot(N, np.ones(len(N)), label=r"$\Omega(1)$")

plt.title(f"Complejidad Mínimo 2")
plt.xlabel("n")
plt.ylabel("Operaciones")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
````

# Análisis A posteriori de la función `Mínimo 2`

## Implementación

La idea principal de este algoritmo es recorrer todo le arreglo de atras hacia adelante de forma recursiva, es decir, se compara el elemnto actual con el minimo encontrado al llamar la misma función, sin embrago en cada llamada va a buscar el minimo del extremo no investigado, esto seguira una y otra vez hasta que se llegue al principio del arreglo.


NOTA:
- Uno de los problemas dentro de la recursión es la cantidad de llamdas, esto quiere decir que si tendría `n` llamadas podría llegar al tope establecido del sistema

A continuación, se muestra su implementación


````python
def minimo2(A: int , n: int) -> int:
    if n == 0: # O(1)
        return A[0]
    else:
        aux = minimo2(A, n - 1) # T(n - 1)
        if A[n] > aux: # O (1)
            return aux
        else:
            return A[n]

arr = [1, 2, 3, 4, 5, 7, -23, 40]
print("Resultado: ", minimo2(arr, len(arr) - 1))

size_data = [
    10,
    20,
    30,
    60,
    70,
    80,
    90,
    100,
    120,
    140,
    200,
    300
]


# ==========================================================
# GENERACIÓN DE DATASETS
# ==========================================================

test_data = []

for size in size_data:
    sample = [
        random.randint(-1000000, 1000000)
        for _ in range(size)
    ]
    test_data.append(sample)


# ==========================================================
# MEDICIÓN EXPERIMENTAL
# ==========================================================

times = []

REPETITIONS = 100

for test in test_data:
    start = time.perf_counter()
    for _ in range(REPETITIONS):
        minimo2(test, len(test) - 1)
    end = time.perf_counter()
    # promedio en milisegundos
    curr_time = (
        (end - start) / REPETITIONS
    ) * 1000

    print(f"n={len(test)} -> "f"{curr_time:.8f} ms")
    times.append(curr_time)


# ==========================================================
# COMPLEJIDAD TEÓRICA Θ(n)
# ==========================================================

theoretical_raw = size_data

# Escalar para comparar tendencias
scale_factor = (
    times[-1] /
    theoretical_raw[-1]
)

theoretical = [
    x * scale_factor
    for x in theoretical_raw
]


# ==========================================================
# RATIO
# ==========================================================

ratio = [
    times[i] / size_data[i]
    for i in range(len(size_data))
]


# ==========================================================
# GRÁFICAS
# ==========================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(15, 5)
)

fig.suptitle(
    "Análisis A Posteriori — minimo2",
    fontsize=14,
    y=1.02
)

# ----------------------------------------------------------
# 1. Tiempo real vs teórico
# ----------------------------------------------------------

axes[0].plot(
    size_data,
    times,
    "o-",
    color="#E24B4A",
    linewidth=2,
    label="Tiempo real"
)

axes[0].plot(
    size_data,
    theoretical,
    "--",
    color="#378ADD",
    linewidth=2,
    label=r"$\Theta(n)$ teórico"
)

axes[0].set_xscale("log")
axes[0].set_yscale("log")

axes[0].set_title(
    "Tiempo real vs Teórico"
)

axes[0].set_xlabel(
    "n (número de elementos)"
)

axes[0].set_ylabel(
    "Tiempo promedio (ms)"
)

axes[0].legend()

axes[0].grid(alpha=0.4)

# ----------------------------------------------------------
# 2. Ratio
# ----------------------------------------------------------

axes[1].plot(
    size_data,
    ratio,
    "o-",
    color="#1D9E75",
    linewidth=2
)

axes[1].set_xscale("log")

axes[1].set_title(
    r"Ratio $T(n)/n$"
)

axes[1].set_xlabel(
    "n (número de elementos)"
)

axes[1].set_ylabel(
    r"$T(n)/n$"
)

axes[1].grid(alpha=0.4)

# ==========================================================

plt.tight_layout()

plt.savefig(
    "IMG/minimo2_aposteriori.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "\nGráfica guardada como "
    "minimo1_aposteriori.png"
)
````

# Análisis A priori de la función `Mínimo 3`

Con base a al definición de la función `Mínimo 3`, se analizara para obtener su compllejidad en tiempo. 


A continuación, se muestra el código de la función y como comentario se agrega su complejidad en cada declaración.


```python

def minimo3(A: int , ini: int, fin: int) -> int:
    if ini == fin: # O(1)
        return A[ini]
    else:
        mid = (ini + fin) // 2 # O(1)
        if minimo3(A, ini, mid) < minimo3(A, mid + 1, fin): # T(n/2) + T(n/2)
            return minimo3(A, ini, mid) # T(n/2)
        else:
            return minimo3(A, mid + 1, fin) # T(n/2)
```

La función `Mínimo 3` es muy similar a la función `Mínimo 1`, sin embargo, aquí se hacen múliples llamadas con la misma configuraicón. Lo destaclabe es que en el peor de los caso es cuando se evalua el segmento `else`, lo que implicaría un `3T(n/2)`, al construir al recurrencia tenemos:

$$

T(n) = 3T(n/2) + C

$$

Y vemos que es igual a la forma $T(n) = aT(n/b) + f(n)$, donde podemos aplicar el Teorema maestro. Entonces $a = 3$ , $b = 2$ y $f(n) = C$.

$$

n^{log_{b}(a)}, \quad n^{log_{2}(3)} = n  \\

y \\

f(n) = O(C),
$$

Por lo tanto, se aplica el caso 1 del Teorema Maestro. En otras palabras, $C$ al no depender de n, no crece y $n^{log_{2}(3) - \epsilon}$ con $\epsilon = 1$, tenemos que:

$$

n^{log_{2}(3) - 1} = n^{1.585 - 1} = n^{0.585} \\

O(C) = 1 \in \Theta(n^{0.585})

$$

Y tenemos una complejidad de $\Theta(n^{1.585})$

# Análisis A posteriori de la función `Mínimo 3`

## Implementación

Esta función busca el minimo por los dos lados, es decir, primero se encontra la mitad y se invoca a si misma para encontrar el minimo entre los rangos superiores y menores, pero lo que la hace costosa, es la cantidad de invocaciones que hace.

````python
import time
import random
import numpy as np
import matplotlib.pyplot as plt

def minimo3(A: int , ini: int, fin: int) -> int:
    if ini == fin:
        return A[ini]
    else:
        mid = (ini + fin) // 2
        if minimo3(A, ini, mid) < minimo3(A, mid + 1, fin):
            return minimo3(A, ini, mid)
        else:
            return minimo3(A, mid + 1, fin)

arr = [1, 2, 3, 4, 5, 7, -23, 40]
print("Resultado: ", minimo3(arr, 0, len(arr) - 1))

size_data = [
    10,
    100,
    1000,
    5000,
    10000,
    50000,
]


# ==========================================================
# GENERACIÓN DE DATASETS
# ==========================================================

test_data = []

for size in size_data:
    sample = [
        random.randint(-1000000, 1000000)
        for _ in range(size)
    ]
    test_data.append(sample)


# ==========================================================
# MEDICIÓN EXPERIMENTAL
# ==========================================================

times = []

REPETITIONS = 100

for test in test_data:
    start = time.perf_counter()
    for _ in range(REPETITIONS):
        minimo3(
            test,
            0,
            len(test) - 1
        )
    end = time.perf_counter()
    # promedio en milisegundos
    curr_time = (
        (end - start) / REPETITIONS
    ) * 1000

    print(f"n={len(test)} -> "f"{curr_time:.8f} ms")
    times.append(curr_time)


# ==========================================================
# COMPLEJIDAD TEÓRICA Θ(n)
# ==========================================================

theoretical_raw = size_data

# Escalar para comparar tendencias
scale_factor = (
    times[-1] /
    theoretical_raw[-1]
)

theoretical = [
    x * scale_factor
    for x in theoretical_raw
]


# ==========================================================
# RATIO
# ==========================================================

ratio = [
    times[i] / size_data[i]
    for i in range(len(size_data))
]


# ==========================================================
# GRÁFICAS
# ==========================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(15, 5)
)

fig.suptitle(
    "Análisis A Posteriori — minimo3",
    fontsize=14,
    y=1.02
)

# ----------------------------------------------------------
# 1. Tiempo real vs teórico
# ----------------------------------------------------------

axes[0].plot(
    size_data,
    times,
    "o-",
    color="#E24B4A",
    linewidth=2,
    label="Tiempo real"
)

axes[0].plot(
    size_data,
    theoretical,
    "--",
    color="#378ADD",
    linewidth=2,
    label=r"$\Theta(n)$ teórico"
)

axes[0].set_xscale("log")
axes[0].set_yscale("log")

axes[0].set_title(
    "Tiempo real vs Teórico"
)

axes[0].set_xlabel(
    "n (número de elementos)"
)

axes[0].set_ylabel(
    "Tiempo promedio (ms)"
)

axes[0].legend()

axes[0].grid(alpha=0.4)

# ----------------------------------------------------------
# 2. Ratio
# ----------------------------------------------------------

axes[1].plot(
    size_data,
    ratio,
    "o-",
    color="#1D9E75",
    linewidth=2
)

axes[1].set_xscale("log")

axes[1].set_title(
    r"Ratio $T(n)/n$"
)

axes[1].set_xlabel(
    "n (número de elementos)"
)

axes[1].set_ylabel(
    r"$T(n)/n$"
)

axes[1].grid(alpha=0.4)

# ==========================================================

plt.tight_layout()

plt.savefig(
    "IMG/minimo3_aposteriori.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "\nGráfica guardada como "
    "minimo3_aposteriori.png"
)
````

## Implementación

La función `mcd`es corresponde al algoritmo de restas sucesiva para encontrar el mínimo común divisor. Cada invoación uno de los elemento disminuye en función del otro.

A continuación, se muestra su implementación


````python
import time
import random
import numpy as np
import matplotlib.pyplot as plt

def mcd(x: int, y: int) -> int:
    if x == y:
        return x
    elif x > y:
        return mcd( x - y, y)
    else:
        return mcd(x, y -x)

print("MCD: ", mcd(4, 6))

size_data = [
    10,
    50,
    100,
    500,
    1000,
]


# ==========================================================
# MEDICIÓN EXPERIMENTAL
# ==========================================================

times = []

REPETITIONS = 1000

for n in size_data:

    # Peor caso aproximado:
    # mcd(n,1)
    x = n
    y = 1

    start = time.perf_counter()

    for _ in range(REPETITIONS):
        mcd(x, y)

    end = time.perf_counter()

    # promedio en milisegundos
    curr_time = (
        (end - start) / REPETITIONS
    ) * 1000

    print(
        f"n={n} -> "
        f"{curr_time:.8f} ms"
    )

    times.append(curr_time)


# ==========================================================
# COMPLEJIDAD TEÓRICA Θ(n)
# ==========================================================

theoretical_raw = size_data

# Escalamiento para comparar tendencia
scale_factor = (
    times[-1] /
    theoretical_raw[-1]
)

theoretical = [
    x * scale_factor
    for x in theoretical_raw
]


# ==========================================================
# RATIO
# ==========================================================

ratio = [
    times[i] / size_data[i]
    for i in range(len(size_data))
]


# ==========================================================
# GRÁFICAS
# ==========================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(15, 5)
)

fig.suptitle(
    "Análisis A Posteriori — MCD por Restas",
    fontsize=14,
    y=1.02
)

# ----------------------------------------------------------
# 1. Tiempo real vs Teórico
# ----------------------------------------------------------

axes[0].plot(
    size_data,
    times,
    "o-",
    color="#E24B4A",
    linewidth=2,
    label="Tiempo real"
)

axes[0].plot(
    size_data,
    theoretical,
    "--",
    color="#378ADD",
    linewidth=2,
    label=r"$O(n)$ teórico"
)

axes[0].set_xscale("log")
axes[0].set_yscale("log")

axes[0].set_title(
    "Tiempo real vs Teórico"
)

axes[0].set_xlabel(
    "n"
)

axes[0].set_ylabel(
    "Tiempo promedio (ms)"
)

axes[0].legend()

axes[0].grid(alpha=0.4)

# ----------------------------------------------------------
# 2. Ratio
# ----------------------------------------------------------

axes[1].plot(
    size_data,
    ratio,
    "o-",
    color="#1D9E75",
    linewidth=2
)

axes[1].set_xscale("log")

axes[1].set_title(
    r"Ratio $T(n)/n$"
)

axes[1].set_xlabel(
    "n"
)

axes[1].set_ylabel(
    r"$T(n)/n$"
)

axes[1].grid(alpha=0.4)

# ==========================================================

plt.tight_layout()

plt.savefig(
    "IMG/mcd_aposteriori.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "\nGráfica guardada como "
    "mcd_aposteriori.png"
)
````
